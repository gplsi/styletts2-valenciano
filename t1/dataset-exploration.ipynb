## Exploración LJSpeech

```python
from datasets import load_dataset, Audio, Dataset

# El dataset original es "keithito/lj_speech" pero tiene un formato no soportado por las nuevas versiones de datasets
# por lo que usamos este que tiene el mismo contenido pero en un formato compatible
dataset: Dataset = load_dataset("MikhailT/lj-speech", split="full[:20]")
```

```python
dataset = dataset.cast_column("audio", Audio(sampling_rate=22050))

print(dataset)
```

```python
from IPython.display import Audio as IPyAudio

row = dataset[0]
audio_path = row["audio"]
print(f"Spoken text: {row['spoken_text']}")
print(f"Normalized text: {row['normalized_text']}")
IPyAudio(audio_path["array"], rate=audio_path["sampling_rate"])
```

## Exploración LibriTTS

```python
dataset: Dataset = load_dataset("mythicinfinity/libritts", "dev", split="dev.clean")
```

```python
dataset = dataset.cast_column("audio", Audio(sampling_rate=24000)) # LibriTTS usa 24 kHz como se puede ver en https://www.openslr.org/60/
print(dataset)
```

```python
from IPython.display import Audio as IPyAudio

row = dataset[0]
audio_path = row["audio"]
print(f"Spoken text: {row['text_original']}")
print(f"Normalized text: {row['text_normalized']}")
print(f"Speaker ID: {row['speaker_id']}")
IPyAudio(audio_path["array"], rate=audio_path["sampling_rate"])
```

### Exploración de speakers

```python
speaker_ids = dataset.unique("speaker_id")
print(f"Número de speakers: {len(speaker_ids)} (al menos en dev split)")
print(sorted(speaker_ids))  # muestra todos los speakers
```

```python
# Probamos a reproducir otro speaker
speaker_id = "1462"

# Filtramos el dataset para este speaker
speaker_dataset = dataset.filter(lambda x: x["speaker_id"] == speaker_id)

row = speaker_dataset[0]
audio_path = row["audio"]
print(f"Spoken text: {row['text_original']}")
print(f"Normalized text: {row['text_normalized']}")
print(f"Speaker ID: {row['speaker_id']}")
IPyAudio(audio_path["array"], rate=audio_path["sampling_rate"])
```
